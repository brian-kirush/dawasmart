# DawaSmart forecasting evaluation

This reproducible example evaluates one clearly synthetic chronic-medicine series. It uses the project train, validation and test dates and reports the metrics; it does not write sales or forecasts to Supabase.

```python
from datetime import date
import pandas as pd

from forecasting.data import generate_synthetic_history
from forecasting.evaluation import evaluate_series
from forecasting.features import daily_series, prepare_series

medicine = {
    "id": "synthetic-evaluation-medicine",
    "generic_name": "Synthetic chronic medicine",
    "therapeutic_class": "Antidiabetic",
    "demand_pattern": "chronic",
}
medicines = [medicine]
records = generate_synthetic_history(
    medicines,
    start_date=date(2022, 1, 1),
    end_date=date(2024, 12, 31),
)
series = daily_series(
    records,
    medicine["id"],
    start=date(2022, 1, 1),
    end=date(2024, 12, 31),
)
series, frequency = prepare_series(series, decision_end=date(2024, 6, 30))
frequency, len(series), series.head()
```

```python
evaluation = evaluate_series(series, medicine, medicines)
validation_comparison = pd.DataFrame.from_dict(
    evaluation["validation_by_model"],
    orient="index",
).rename_axis("model")
validation_comparison["selected"] = validation_comparison.index == evaluation["selected_model"]
validation_comparison
```

The selected model's untouched test metrics and stock simulation are available in `evaluation['test']` and `evaluation['simulation']`. The 15% sMAPE improvement is a target, not a result assumed in advance. Use measured test metrics when writing the project report. Synthetic results are not evidence of real-pharmacy performance.
